# Iris data analysis

Explore the iris dataset from scikit-learn: 150 flowers, 4 measurements (sepal and petal, in cm) and the species (`target`: setosa, versicolor or virginica).


```python
import matplotlib.pyplot as plt
from sklearn.datasets import load_diabetes
```

```python
print("Welcome to the Diabetes tutorial of the HIDS Python Programming course!")
```

## 1. Load the data


```python
data = load_diabetes(as_frame=True)
df = data.frame
df.head()
```

```python
df.shape
```

## 2. Descriptive statistics


```python
df.describe()
```

```python
df.isna().sum()
```

## 3. Distribution of the target


```python
df["target"].plot.hist(bins=30, title="Disease progression")
plt.xlabel("target")
plt.show()
```

## 4. Correlation with the target


```python
corr = df.corr()["target"].drop("target").sort_values()
corr.plot.barh(title="Correlation with target")
plt.show()
```

## 5. Relationship with the most correlated feature


```python
best = corr.abs().idxmax()
df.plot.scatter(x=best, y="target", alpha=0.6, title=f"target vs {best}")
plt.show()
```
